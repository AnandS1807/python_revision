```python
class Shape(ABC):
    @abstractmethod
    def area(self):
        """subclasses must defined this method"""
        pass
        
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        print(id(self), "address of self")
    def area(self):
        return 3.14*(self.radius**2)
```

```python
o1 = Circle(5)
print(o1.area())
print(id(o1), "address of object o1")

o2 = Circle(6)
print(o2.area())
print(id(o2), "address of object o2")
```

```python
def func(max):
    count = 1;
    while count<max:
        yield(count)
        count+=1
```

```python
obj = func(5)
for i in obj:
    print(i)
```

```python
import math

class Twodimgeometry:

    def __init__(self):
        print("2D Geometry Toolkit Ready")

    # 1. Distance between two points
    def distance(self, x1, y1, x2, y2):
        return math.sqrt((x2-x1)**2 + (y2-y1)**2)

    # 2. Midpoint of two points
    def midpoint(self, x1, y1, x2, y2):
        return ((x1+x2)/2, (y1+y2)/2)

    # 3. Slope of line
    def slope(self, x1, y1, x2, y2):
        if x2 == x1:
            return None
        return (y2-y1)/(x2-x1)

    # 4. Equation of line from two points
    def line_from_points(self, x1, y1, x2, y2):
        a = y2 - y1
        b = x1 - x2
        c = x2*y1 - x1*y2
        return a, b, c

    # 5. Distance between point and line
    def point_line_distance(self, x, y, a, b, c):
        return abs(a*x + b*y + c) / math.sqrt(a*a + b*b)

    # 6. Distance between parallel lines
    def parallel_line_distance(self, a, b, c1, c2):
        return abs(c1-c2)/math.sqrt(a*a+b*b)

    # 7. Intersection of two lines
    def line_intersection(self, a1, b1, c1, a2, b2, c2):
        det = a1*b2 - a2*b1
        if det == 0:
            return "Parallel lines"
        x = (b1*c2 - b2*c1)/det
        y = (c1*a2 - c2*a1)/det
        return (x,y)

    # 8. Angle between two lines
    def angle_between_lines(self, m1, m2):
        if (1 + m1*m2) == 0:
            return 90
        angle = abs((m2-m1)/(1+m1*m2))
        return math.degrees(math.atan(angle))

    # 9. Area of triangle using coordinates
    def triangle_area(self, x1,y1,x2,y2,x3,y3):
        return abs(x1*(y2-y3)+x2*(y3-y1)+x3*(y1-y2))/2

    # 10. Centroid of triangle
    def triangle_centroid(self, x1,y1,x2,y2,x3,y3):
        return ((x1+x2+x3)/3,(y1+y2+y3)/3)

    # 11. Equation of circle
    def circle_equation(self, h,k,r):
        return f"(x-{h})^2 + (y-{k})^2 = {r*r}"

    # 12. Distance of point from circle center
    def point_from_circle_center(self,x,y,h,k):
        return math.sqrt((x-h)**2 + (y-k)**2)

    # 13. Check if point lies inside circle
    def point_inside_circle(self,x,y,h,k,r):
        d = self.point_from_circle_center(x,y,h,k)
        return d < r

    # 14. Tangent to circle at point
    def tangent_to_circle(self,h,k,r,x1,y1):
        return f"({x1}-{h})(x-{h}) + ({y1}-{k})(y-{k}) = {r*r}"

    # 15. Polygon area (Shoelace formula)
    def polygon_area(self,points):
        n=len(points)
        area=0
        for i in range(n):
            x1,y1=points[i]
            x2,y2=points[(i+1)%n]
            area += x1*y2 - x2*y1
        return abs(area)/2
```

```python
import logging
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

# 1. Setup Logging (Architects love logs)
logging.basicConfig(level=logging.INFO)

def process_docs_to_faiss(file_path, index_path="faiss_index"):
    try:
        # 2. Efficient Loading
        loader = PyPDFLoader(file_path)
        documents = loader.load()
        
        # 3. Smart Chunking
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=100,
            length_function=len
        )
        chunks = text_splitter.split_documents(documents)
        
        # 4. Local Embeddings (Privacy-focused)
        embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
        
        # 5. Create & Save Index
        vector_db = FAISS.from_documents(chunks, embeddings)
        vector_db.save_local(index_path)
        logging.info(f"Successfully indexed {len(chunks)} chunks.")
        
    except Exception as e:
        logging.error(f"Pipeline failed: {str(e)}")

# Example call
# process_docs_to_faiss("large_manual.pdf")
```

```python
process_docs_to_faiss("sample_a_cruv (1).pdf")
```
